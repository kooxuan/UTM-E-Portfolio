```
#question 1b
import numpy as np

def normalize(x):
    fac = abs(x).max()
    x_n = x/x.max()
    return fac, x_n

# Initial vector
x = np.array([0, 1, 1])

# Matrix A
A = np.array([[0, 11, -5], [-2, 17, -7], [-4, 26, -10]])

# Perform the Power Method
lambda_old=0
for i in range(8):
  x = np.dot(A, x)
  lambda_new, x = normalize(x)
  if np.abs(lambda_new-lambda_old)<0.005:
    break
  lambda_old=lambda_new
print("Eigenvalue: ", round(lambda_new, 2))
print("Eigenvector: ", x)
```

```
#question 2i
import numpy as np

def newton_interpolation(x, y, x_val):
    """
    Perform Newton's interpolation for given data points (x, y) and estimate the value at x_val.
    """
    n = len(x)
    divided_diff = np.zeros((n, n))
    divided_diff[:,0] = y

    for j in range(1, n):
        for i in range(n-j):
            divided_diff[i, j] = (divided_diff[i+1, j-1] - divided_diff[i, j-1]) / (x[i+j] - x[i])

    # Calculating the interpolation value at x_val
    estimated_val = divided_diff[0, 0]
    product_term = 1.0
    for i in range(1, n):
        product_term *= (x_val - x[i-1])
        estimated_val += divided_diff[0, i] * product_term

    return estimated_val

# Given data points
T = np.array([5, 20, 30, 50, 55], dtype=float)
mu = np.array([0.0800, 0.0150, 0.0090, 0.0060, 0.0055], dtype=float)

# Value to estimate
T_val = 40

# Perform Newton's Interpolation
mu_estimated = newton_interpolation(T, mu, T_val)
print(f"T={T_val}")
print(f"Estimate viscosity= {mu_estimated:.4f}")
```

```
#question 3
import numpy as np

# Given data points
labour_cost = np.array([10, 16, 25, 40, 60], dtype=float)
profit = np.array([94, 118, 147, 180, 230], dtype=float)

# Perform least squares fitting to find a and b
A = np.vstack([labour_cost, np.ones(len(labour_cost))]).T
b, a = np.linalg.lstsq(A, profit, rcond=None)[0]

print(f"Least squares values:")
print(f"a: {a:.4f}")
print(f"b: {b:.4f}")

# Function to calculate profit
def calculate_profit(labour_cost, a, b):
    return a + b * labour_cost

# b) Find the profit if the labour cost is 210
labour_cost_210 = 210
predicted_profit_210 = calculate_profit(labour_cost_210, a, b)
print(f"\nThe estimated profit is RM{predicted_profit_210:.2f}")
```

```
#question 4
import sympy as sp

# Define the function
x = sp.symbols('x')
f_expr = 2*x**3 + x**2 - 4

# Define the function f(x)
def f(x):
    return 2*x**3 + x**2 - 4

# Define h
h = 0.5
x0 = 2.5

# Central three-point formula for f'(2.5) and f''(2.5)
f_prime_3_point = (f(x0 + h) - f(x0 - h)) / (2 * h)
f_double_prime_3_point = (f(x0 + h) - 2 * f(x0) + f(x0 - h)) / (h**2)

# Central five-point formula for f'(2.5) and f''(2.5)
f_prime_5_point = (-f(x0 + 2*h) + 8*f(x0 + h) - 8*f(x0 - h) + f(x0 - 2*h)) / (12 * h)
f_double_prime_5_point = (-f(x0 + 2*h) + 16*f(x0 + h) - 30*f(x0) + 16*f(x0 - h) - f(x0 - 2*h)) / (12 * h**2)

print(f"Estimates using central formulas:")
print(f"Three-point formula:\n f'(2.5) ≈ {f_prime_3_point}\n f''(2.5) ≈ {f_double_prime_3_point}")
print(f"Five-point formula:\n f'(2.5) ≈ {f_prime_5_point}\n f''(2.5) ≈ {f_double_prime_5_point}")

# Compute the exact values of f'(2.5) and f''(2.5)
f_prime_expr = sp.diff(f_expr, x)
f_double_prime_expr = sp.diff(f_prime_expr, x)

# Substitute x = 2.5
f_prime_2_5 = f_prime_expr.subs(x, 2.5)
f_double_prime_2_5 = f_double_prime_expr.subs(x, 2.5)

print(f"\nExact values:\n f'(2.5) = {f_prime_2_5}\n f''(2.5) = {f_double_prime_2_5}")

# Calculate errors
error_f_prime_3_point = abs(f_prime_3_point - f_prime_2_5)
error_f_prime_5_point = abs(f_prime_5_point - f_prime_2_5)
error_f_double_prime_3_point = abs(f_double_prime_3_point - f_double_prime_2_5)
error_f_double_prime_5_point = abs(f_double_prime_5_point - f_double_prime_2_5)

print(f"\nErrors:")
print(f"Error in f'(2.5) using three-point formula: {error_f_prime_3_point}")
print(f"Error in f'(2.5) using five-point formula: {error_f_prime_5_point}")
print(f"Error in f''(2.5) using three-point formula: {error_f_double_prime_3_point}")
print(f"Error in f''(2.5) using five-point formula: {error_f_double_prime_5_point}")
```

```
#question 5a
import numpy as np

# Define the function
def f(x):
    return 2 / (1 + 5 * x**2)

# Define the interval and number of subintervals
a = 0
b = 3
N = 6
h = (b - a) / N

# Generate the x values
x = np.linspace(a, b, N+1)

# Apply the Trapezoidal rule
integral = (h / 2) * (f(x[0]) + 2 * np.sum(f(x[1:N])) + f(x[N]))

print(f"Approximate value of the integral using the Trapezoidal rule with N={N} is: {integral:.4f}")
```

```
#question 5d
import numpy as np

# Define the function
def f(x):
    return 2 / (1 + 5 * x**2)

# Define the Romberg integration function
def romberg_integration(f, a, b, epsilon):
    R = [[0.5 * (b - a) * (f(a) + f(b))]]  # R[0][0]
    n = 1
    k = 0

    while True:
        h = (b - a) / 2**n
        summation = sum(f(a + (2 * i - 1) * h) for i in range(1, 2**(n-1) + 1))
        row_k1 = [0.5 * R[k][0] + summation * h]

        for j in range(1, k + 2):
            row_k1.append(row_k1[j-1] + (row_k1[j-1] - R[k][j-1]) / (4**j - 1))

        if abs(row_k1[-1] - R[k][-1]) < epsilon:
            return row_k1[-1]

        R.append(row_k1)
        n += 1
        k += 1

# Define the interval and epsilon
a = 0
b = 3
epsilon = 0.005

# Apply Romberg integration
integral = romberg_integration(f, a, b, epsilon)

print(f"Approximate value of the integral using Romberg integration with epsilon={epsilon} is: {integral:.4f}")
```

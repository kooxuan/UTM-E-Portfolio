```python
#Q3
import math

# Defining Function
def f(x):
    return 3 * x - math.exp(-x)

# Implementing Bisection Method
def bisection(x0, x1, tol):
    step = 1
    print('\n\n*** BISECTION METHOD IMPLEMENTATION ***')
    condition = True

    while condition:
        x2 = (x0 + x1) / 2
        print(f"Iteration-{step}, x2 = {x2:.6f} and f(x2) = {f(x2):.6f}")

        if f(x0) * f(x2) < 0:
            x1 = x2
        else:
            x0 = x2

        step += 1
        condition = abs(f(x2)) > tol

    print(f"\nRequired Root is : {x2:.3f}")

# Input Section
x0 = float(input('First Guess: '))
x1 = float(input('Second Guess: '))
tol = float(input('Tolerable Error: '))

# Checking Correctness of initial guess values and bisecting
if f(x0) * f(x1) > 0.0:
    print('Given guess values do not bracket the root.')
    print('Try Again with different guess values.')
else:
    bisection(x0, x1, tol)
```

```python
#Q4a
import numpy as np

# Defining Function
def f(x):
    return np.cos(x) + 2 * np.sin(x) + x**2

# Implementing Secant Method
def secant(x0, x1, e, N=100):
    print('\n\n* SECANT METHOD IMPLEMENTATION *')
    step = 1
    condition = True

    while condition and step <= N:
        if f(x0) == f(x1):
            print('Divide by zero error!')
            break

        x2 = x0 - (x1 - x0) * f(x0) / (f(x1) - f(x0))
        print('Iteration-%d, x2 = %0.6f and f(x2) = %0.6f' % (step, x2, f(x2)))

        x0, x1 = x1, x2
        step += 1
        condition = abs(f(x2)) > e

    if step > N:
        print('Maximum iterations reached without convergence.')
    else:
        print('\nRequired root is: %0.2f' % x2)

# Input Section
x0 = float(input('Enter First Guess: '))
x1 = float(input('Enter Second Guess: '))
e = float(input('Tolerable Error: '))
N = int(input('Maximum Iterations: '))

# Starting Secant Method
secant(x0, x1, e, N)
```

```python








#Q4b
import numpy as np

# Defining Function
def f(x):
    return np.cos(x) + 2 * np.sin(x) + x**2

# Defining derivative of function
def g(x):
    return -np.sin(x)+2*np.cos(x)+2*x

# Implementing Newton-Raphson Method
def newtonRaphson(x0, e, N):
    print('\n\n* NEWTON RAPHSON METHOD IMPLEMENTATION *')
    step = 1
    condition = True

    while condition and step <= N:
        if g(x0) == 0.0:
            print('Divide by zero error!')
            break

        x1 = x0 - f(x0) / g(x0)
        print('Iteration-%d, x1 = %0.6f and f(x1) = %0.6f' % (step, x1, f(x1)))

        x0 = x1
        step += 1
        condition = abs(f(x1)) > e

    if step > N:
        print('Maximum iterations reached without convergence.')
    else:
        print('\nRequired root is: %0.2f' % x1)

# Input Section
x0 = float(input('Enter Guess: '))
e = float(input('Tolerable Error: '))
N = int(input('Maximum Iterations: '))

# Starting Newton-Raphson Method
newtonRaphson(x0, e, N)
```
